```python
df
```

```python
XDe = df[['Edad']]
print(XDe)
XDe=XDe.dropna() #evitar celdas vacias
XDe=np.array(XDe)
```

Se calcula la media y la desviacion estandar de la edad

```python
print("La media de la edad es :", np.mean(XDe))
print("La desviacion estandar de la variable es :", np.std(XDe))
print("la edad minima es:", np.mean(XDe)-2*np.std(XDe))
print("la edad maxima es:", np.mean(XDe)+2*np.std(XDe))
```

**Histograma de frecuencias**

```python
import matplotlib.pyplot as plt
plt.hist(XDe,bins=20)
plt.show()
```

**Calcular el valor de pertenencia de una persona de 60 años **

```python
VP=np.exp(-0.5*(((49.12-np.mean(XDe))/np.std(XDe))**2))
print("El valor de pertenencia es :", VP)
```
